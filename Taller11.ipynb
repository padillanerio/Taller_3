````python
class Nodo:
  def __init__(self, dato):
    self.dato = dato
    self.siguiente = None
    self.anterior = None
class Transacciones:
  def __init__(self,fecha,monto,tipo):
    self.fecha=fecha
    self.monto=monto
    self.tipo=tipo
````

````python
class ListaTransacciones:
  def __init__(self):
    self.primero=None
    self.ultimo=None
  def insertar(self,fecha,monto,tipo):
    transaccion=Transacciones(fecha,monto,tipo)
    nuevo_nodo=Nodo(transaccion)
    if self.primero is None:
      self.primero=nuevo_nodo
      self.ultimo=nuevo_nodo
    else:
      self.ultimo.siguiente=nuevo_nodo
      nuevo_nodo.anterior=self.ultimo
      self.ultimo=nuevo_nodo
      nuevo_nodo.siguiente=None
  def calcularSaldo(self):
    if self.primero is None:
      return 0
    else:
      saldo=0
      actual=self.primero
      while actual is not None:
        if actual.dato.tipo=="ingreso":
          saldo+=actual.dato.monto
        else:
          saldo-=actual.dato.monto
        actual=actual.siguiente

  def revertir_ultima(self):
    if self.ultimo is None:
        print("No hay transacciones para revertir")
        return

    nodo = self.ultimo
    self.ultimo = nodo.anterior
    if self.ultimo is None:
        # Era el único nodo
        self.primero = None
    else:
        self.ultimo.siguiente = None


    nodo.anterior = None
    nodo.siguiente = None
````
