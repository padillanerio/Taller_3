````python
class Nodo:
  def __init__(self, dato):
    self.dato = dato
    self.ant = None
    self.sig = None
class Contacto:
  def __init__(self, nombre, telefono,email):
    self.nombre = nombre
    self.telefono = telefono
    self.email=email
````

````python
class ListaContactos:
  def __init__(self):
    self.primero = None
    self.ultimo = None
  def agregar(self, contacto):
    nuevo = Nodo(contacto)
    if self.primero is None:
      self.primero = nuevo
      self.ultimo = nuevo
    else:
      self.ultimo.sig = nuevo
      nuevo.ant = self.ultimo
      self.ultimo = nuevo
  def eliminar(self, nombre):
    actual = self.primero
    while actual is not None:
      if actual.dato.nombre == nombre:
        if actual == self.primero:
          self.primero = actual.sig
        if actual == self.ultimo:
          self.ultimo=actual.ant
        else:
          actual.ant.sig = actual.sig
          actual.sig.ant = actual.ant
        return
      actual = actual.sig
  def buscar(self, nombre):
    actual = self.primero
    while actual is not None:
      if actual.dato.nombre == nombre:
        return actual.dato.nombre, actual.dato.telefono, actual.dato.email
      actual = actual.sig
    return None

  def ordenar(self):
        if self.primero is None:
            return

        cambiado = True
        while cambiado:
            cambiado = False
            actual = self.primero

            while actual.sig is not None:
                if actual.dato.nombre > actual.sig.dato.nombre:
                    actual.dato, actual.sig.dato = actual.sig.dato, actual.dato
                    cambiado = True
                actual = actual.sig
    def buscar_por_telefono(self, telefono):
        actual = self.primero
        while actual is not None:
            if actual.dato.telefono == telefono:
                return actual.dato.nombre, actual.dato.telefono, actual.dato.email
            actual = actual.sig
        return None
````
