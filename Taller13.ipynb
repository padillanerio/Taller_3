````python
# Nodo asumido:
# class Node:
#Asumo que conocemos y tenemos definido un ptr y last
def es_circular(head, last):
    if head is None:
        return False
    return last.next == head
def convertir_a_circular(head, last):
    if head is None:
        return head
    last.next = head
    return head
def convertir_a_lineal(last):
    if last is None:
        return
    last.next = None
````
