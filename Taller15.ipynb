````python
#se asume que las fechas son un int
class Nodo:
  def __init__(self, dato):
    self.dato = dato
    self.sig = None

class Logs:
  def __init__(self,IP,fecha,Recurso):
    self.IP = IP
    self.fecha = fecha
    self.Recurso = Recurso
class ListaLogs:
  def __init__(self):
    self.primero = None
    self.ultimo = None
  def insertar(self,dato):
    nuevo = Nodo(dato)
    if self.primero == None:
      self.primero = nuevo
      self.ultimo = nuevo
    else:
      self.ultimo.sig = nuevo
      self.ultimo = nuevo
  def contarIP(self):
    aux = self.primero
    listaIP = []
    while aux != None:
      if aux.dato.IP not in listaIP:
        listaIP.append(aux.dato.IP)
      aux = aux.sig
  def LogsporFechas(self,fecha1,fecha2):
    aux = self.primero
    actual = aux.dato.fecha
    listaFechas = []
    ListaRango = []
    while aux != None:
      listaFechas.append(aux.dato.fecha)
      aux = aux.sig
    for fecha in listaFechas:
      if fecha >= fecha1 and fecha <= fecha2:
        ListaRango.append(fecha)
    return ListaRango
  def LogsPorIP(self):
    aux = self.primero
    listaIP = []
    while aux != None:
      if aux.dato.IP not in listaIP:
        listaIP.append(aux.dato.IP)
      aux = aux.sig
    return len(listaIP)
  def IPmasFrecuentes(self):
    IPSCONT = {}
    aux = self.primero
    while aux != None:
      ip = aux.dato.IP
      IPSCONT[ip] = IPSCONT.get(ip, 0) + 1
      aux = aux.sig

    # Convertir el diccionario a una lista (IP, count) para poder ordenar
    IPSORD = []
    for ip, count in IPSCONT.items():
      IPSORD.append((ip, count))

    for i in range(len(IPSORD) - 1):
      for j in range(len(IPSORD) - i - 1):
        if IPSORD[j][1] < IPSORD[j + 1][1]:
          IPSORD[j], IPSORD[j + 1] = IPSORD[j + 1], IPSORD[j]
    return IPSORD
````
