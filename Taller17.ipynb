````python
class Nodo:
  def __init__(self):
    self.dato = None
    self.siguiente = None
    self.anterior=None
class Sensores:
  def __init__(self,Id,Ubicacion,Valor):
    self.Id=Id
    self.Ubicacion=Ubicacion
    self.Valor=Valor
class ListaSensores:
  def __init__(self):
    self.primero=None
    self.ultimo=None
  def anadir(self,Id,Ubicacion,Valor):
    nuevo=Nodo()
    nuevo.dato=Sensores(Id,Ubicacion,Valor)
    if self.primero==None:
      self.primero=nuevo
      self.ultimo=nuevo
    else:
      self.ultimo.siguiente=nuevo
      nuevo.anterior=self.ultimo
      self.ultimo=nuevo

  def calcularMedia(self):
    actual=self.primero
    suma=0
    contador=0
    while actual!=None:
      suma+=actual.dato.Valor
      contador+=1
      actual=actual.siguiente
    return suma/contador
  def actualizar(self,actualizar,infonueva):
    actual=self.primero
    while actual!=None:
      if actual.dato.Id== actualizar.Id:
        actual.dato=infonueva
      actual=actual.siguiente
  def eliminar(self, id_a_eliminar):
    if self.primero is None:
      print("La lista está vacía, no se puede eliminar.")
      return

    actual = self.primero
    while actual is not None:
      if actual.dato.Id == id_a_eliminar:
        # Caso 1: El nodo a eliminar es el primero
        if actual == self.primero:
          self.primero = actual.siguiente
          if self.primero is not None:
            self.primero.anterior = None
          else: # La lista se quedó vacía
            self.ultimo = None
        # Caso 2: El nodo a eliminar es el último
        elif actual == self.ultimo:
          self.ultimo = actual.anterior
          if self.ultimo is not None:
            self.ultimo.siguiente = None
          else:
            self.primero = None
        # Caso 3: El nodo a eliminar está en el medio
        else:
          actual.anterior.siguiente = actual.siguiente
          actual.siguiente.anterior = actual.anterior
        print(f"Sensor con ID {id_a_eliminar} eliminado exitosamente.")
        return
      actual = actual.siguiente
    print(f"No se encontró el sensor con ID {id_a_eliminar}.")
````
