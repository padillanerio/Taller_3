````python
class Tareas:
  def __init__(self,nombre,prioridad,fecha):
    self.nombre = nombre
    self.prioridad = prioridad
    self.fecha = fecha
````

````python
class Nodo:
  def __init__(self,dato):
    self.dato = dato
    self.siguiente = None
    self.anterior = None
````

````python
class ListaTareas:
    def __init__(self):
        self.primero = None
        self.ultimo = None

    def vacia(self):
        return self.primero is None

    def agregar_ordenado(self, nombre, prioridad, fecha):
        nueva_tarea = Tarea(nombre, prioridad, fecha)
        nuevo_nodo = Nodo(nueva_tarea)

        #  1: La lista está vacía
        if self.vacia():
            self.primero = nuevo_nodo
            self.ultimo = nuevo_nodo
            return

        #  2: Insertar al principio (Mayor prioridad que el primero)
        if nueva_tarea.prioridad > self.primero.tarea.prioridad:
            nuevo_nodo.siguiente = self.primero
            self.primero.anterior = nuevo_nodo
            self.primero = nuevo_nodo
            return

        #  3: Insertar al final (Menor o igual prioridad que el último)
        if nueva_tarea.prioridad <= self.ultimo.tarea.prioridad:
            self.ultimo.siguiente = nuevo_nodo
            nuevo_nodo.anterior = self.ultimo
            self.ultimo = nuevo_nodo
            return

        #  4: Insertar en medio
        actual = self.primero
        while actual.siguiente and actual.siguiente.tarea.prioridad >= nueva_tarea.prioridad:
            actual = actual.siguiente


        nuevo_nodo.siguiente = actual.siguiente
        nuevo_nodo.anterior = actual
        if actual.siguiente:
            actual.siguiente.anterior = nuevo_nodo
        actual.siguiente = nuevo_nodo

    def mostrar_lista(self):
        actual = self.primero
        while actual:
            print(f"[{actual.tarea.prioridad}] {actual.tarea.nombre} - {actual.tarea.fecha}")
            actual = actual.siguiente
````
