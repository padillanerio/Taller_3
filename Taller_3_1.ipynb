### Actividad para practicar manejo de Listas Enlazadas


1. Implementación Básica de una Lista Enlazada Simple:

o Desarrolla una clase ListaEnlazadaSimple que permita insertar, modificar,
eliminar y buscar nodos.

o Implementa métodos para recorrer la lista e imprimir sus elementos.

o Realiza pruebas con diferentes tipos de datos (números, cadenas, objetos).

```python
# CLASE NODO
class node:
    def __init__(self, data):
        self.data = data
        self.next = None

class lista:
    def __init__(self):
        self.head = None
        self.tail = None
    
    def agregar(self, data):
        nuevo_nodo = node(data)
        if self.head is None:
            self.head = nuevo_nodo
            self.tail = nuevo_nodo
        else:
            self.tail.next = nuevo_nodo
            self.tail = nuevo_nodo
        
    def buscar(self, data):
        nodo = self.head
        while nodo is not None:
            if nodo.data == data:
                return nodo
            nodo = nodo.next
        return None
    
    def eliminar(self, data):
        nodo = self.buscar(data)
        if nodo is None:
            return "no se encontro la informacion a eliminar"
        
        if nodo == self.head:
            if nodo == self.tail:
                self.head = None
                self.tail = None
            else:
                aux = self.head.next
                self.head.next = None
                self.head = aux
        elif nodo == self.tail:
            aux = self.head
            while aux.next != self.tail:
                aux = aux.next
            
            aux.next = None
            self.tail = aux
        else:
            aux = self.head
            while aux.next != nodo:
                aux = aux.next
            
            aux.next = nodo.next
            nodo.next = None
        
        return "eliminado"
    
    def mostrar(self):
        nodo_lista = self.head
        while nodo_lista is not None:
            print(f"{nodo_lista.data} -> ", end="")
            nodo_lista = nodo_lista.next
        print("None")
    
    def modificar(self, dato_actual, nuevo_dato):
        x = self.buscar(dato_actual)
        if x is None:
            return "no se encontro la informacion a modificar"
        x.data = nuevo_dato
        return "modificado"
```

```python
# PRUEBAS: numeros, cadenas y objetos
print("=== PRUEBAS CON NUMEROS ===")
lista_numeros = lista()
lista_numeros.agregar(10)
lista_numeros.agregar(20)
lista_numeros.agregar(30)
print("Lista despues de agregar:")
lista_numeros.mostrar()
print("Buscar 20:", lista_numeros.buscar(20).data if lista_numeros.buscar(20) else None)
print("Modificar 20 -> 25:", lista_numeros.modificar(20, 25))
lista_numeros.mostrar()
print("Eliminar 10:", lista_numeros.eliminar(10))
lista_numeros.mostrar()

print("\n=== PRUEBAS CON CADENAS ===")
lista_cadenas = lista()
lista_cadenas.agregar("hola")
lista_cadenas.agregar("python")
lista_cadenas.agregar("listas")
print("Lista despues de agregar:")
lista_cadenas.mostrar()
print("Buscar 'python':", lista_cadenas.buscar("python").data if lista_cadenas.buscar("python") else None)
print("Modificar 'python' -> 'estructuras':", lista_cadenas.modificar("python", "estructuras"))
lista_cadenas.mostrar()
print("Eliminar 'hola':", lista_cadenas.eliminar("listas"))
lista_cadenas.mostrar()

print("\n=== PRUEBAS CON OBJETOS ===")
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def __repr__(self):
        return f"Persona(nombre={self.nombre}, edad={self.edad})"

p1 = Persona("Ana", 20)
p2 = Persona("Luis", 25)
p3 = Persona("Marta", 22)

lista_objetos = lista()
lista_objetos.agregar(p1)
lista_objetos.agregar(p2)
lista_objetos.agregar(p3)
print("Lista despues de agregar:")
lista_objetos.mostrar()
print("Buscar p2:", lista_objetos.buscar(p2))

p2_nuevo = Persona("Luis", 26)
print("Modificar p2 -> p2_nuevo:", lista_objetos.modificar(p2, p2_nuevo))
lista_objetos.mostrar()
print("Eliminar p1:", lista_objetos.eliminar(p2_nuevo))
lista_objetos.mostrar()
```
