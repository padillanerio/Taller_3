## Implementación Básica de una Lista Enlazada Circular Doble:

o Desarrolla una clase ListaEnlazadaCircularDoble que permita insertar,
modificar, eliminar y buscar nodos.

o Implementa métodos para recorrer la lista e imprimir sus elementos.

o Realiza pruebas con diferentes tipos de datos (números, cadenas, objetos).

```python
class node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None
    
class ListaEnlazadaCircularDoble:
    def __init__(self):
        self.head = None
        self.tail = None
    
    def estaVacia(self):
        return self.head is None
    
    def insertar(self,data):
        nodo = node(data)
        if self.estaVacia():
            self.head = nodo
            self.tail = nodo
            nodo.next = nodo
            nodo.prev = nodo
        else:
            self.tail.next = nodo
            nodo.next = self.head
            nodo.prev = self.tail
            self.tail = nodo
            self.head.prev = nodo
    

    def buscar(self, dato):
        if self.estaVacia():
            return f"esta vacia"
        
        puntero = self.head
        while True:
            if puntero.data == dato:
                return puntero
            puntero = puntero.next
            if puntero is self.tail:
                if puntero.data == dato:
                    return puntero
                else:
                    puntero = None
                    return puntero 
        
    def modificar(self, dato_nuevo, dato_viejo):
        if self.estaVacia():
            return
        
        puntero = self.buscar(dato_viejo)

        if puntero != None:
            puntero.data = dato_nuevo
            return f"el valor {dato_viejo} fue modificado por el valor {dato_nuevo}"
        else:
            return f"El valor {dato_viejo} no fue encontrado en la lista"
    
    def eliminar(self, dato_a_eliminar):
        if self.estaVacia():
            return "lista vacia"
        
        cabeza  = self.head
        cola = self.tail
        nodo_a_eliminar = self.buscar(dato_a_eliminar)

        if nodo_a_eliminar is None:
            return "no se encontro el valor a eliminar"

        while True:
            if nodo_a_eliminar is cabeza and nodo_a_eliminar is cola:
                self.head = None
                self.tail = None
            elif nodo_a_eliminar is cabeza:
                self.head = nodo_a_eliminar.next
                self.head.prev = self.tail
                self.tail.next = self.head
            elif nodo_a_eliminar is cola:
                self.tail = nodo_a_eliminar.prev
                self.tail.next = self.head
                self.head.prev = self.tail
            else:
                nodo_a_eliminar.prev.next = nodo_a_eliminar.next
                nodo_a_eliminar.next.prev = nodo_a_eliminar.prev
            return f"el valor {dato_a_eliminar} fue eliminado"
        
    def imprimir(self):
        if self.estaVacia():
            return "lista vacia"
        
        puntero = self.head
        while True:
            print(puntero.data)
            puntero = puntero.next
            if puntero is self.head:
                break
```

```python
# Pruebas de ListaEnlazadaCircularDoble con diferentes tipos de datos
# Se muestran los usos de: estaVacia, insertar, imprimir, buscar, modificar, eliminar

from dataclasses import dataclass

print("=== PRUEBAS CON NUMEROS ===")
lista_num = ListaEnlazadaCircularDoble()
print("Lista vacia al inicio:", lista_num.estaVacia())

lista_num.insertar(10)
lista_num.insertar(20)
lista_num.insertar(30)
print("Contenido despues de insertar 10, 20, 30:")
lista_num.imprimir()

print("Buscar 20 ->", lista_num.buscar(20).data if lista_num.buscar(20) else None)
print(lista_num.modificar(25, 20))
print("Contenido despues de modificar 20 por 25:")
lista_num.imprimir()

print(lista_num.eliminar(10))
print("Contenido despues de eliminar 10:")
lista_num.imprimir()
print("Lista vacia al final:", lista_num.estaVacia())

print("\n=== PRUEBAS CON CADENAS ===")
lista_str = ListaEnlazadaCircularDoble()
print("Lista vacia al inicio:", lista_str.estaVacia())

lista_str.insertar("Ana")
lista_str.insertar("Luis")
lista_str.insertar("Marta")
print("Contenido despues de insertar cadenas:")
lista_str.imprimir()

print("Buscar 'Luis' ->", lista_str.buscar("Luis").data if lista_str.buscar("Luis") else None)
print(lista_str.modificar("Lucia", "Luis"))
print("Contenido despues de modificar 'Luis' por 'Lucia':")
lista_str.imprimir()

print(lista_str.eliminar("Ana"))
print("Contenido despues de eliminar 'Ana':")
lista_str.imprimir()
print("Lista vacia al final:", lista_str.estaVacia())

print("\n=== PRUEBAS CON OBJETOS ===")
@dataclass
class Persona:
    id: int
    nombre: str

lista_obj = ListaEnlazadaCircularDoble()
print("Lista vacia al inicio:", lista_obj.estaVacia())

p1 = Persona(1, "Carlos")
p2 = Persona(2, "Elena")
p3 = Persona(3, "Sofia")

lista_obj.insertar(p1)
lista_obj.insertar(p2)
lista_obj.insertar(p3)
print("Contenido despues de insertar objetos Persona:")
lista_obj.imprimir()

encontrado = lista_obj.buscar(Persona(2, "Elena"))
print("Buscar Persona(2, 'Elena') ->", encontrado.data if encontrado else None)

print(lista_obj.modificar(Persona(2, "Elena M."), Persona(2, "Elena")))
print("Contenido despues de modificar objeto Persona:")
lista_obj.imprimir()

print(lista_obj.eliminar(Persona(1, "Carlos")))
print("Contenido despues de eliminar Persona(1, 'Carlos'):")
lista_obj.imprimir()
print("Lista vacia al final:", lista_obj.estaVacia())
```
