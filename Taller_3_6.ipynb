Optimización de Búsquedas en Listas Enlazadas: 

Dada una lista enlazada de enteros, implementa una función que busque un número n más rápido utilizando un algoritmo de salto. 

Analiza el rendimiento de este algoritmo comparado con una búsqueda lineal. 

````python
class nodo:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None
    
class lista_doble_enlazada:
    def __init__(self):
        self.head = None
        self.tail = None
        self.mitad = None
        self.size = 0  

    def agregar(self, data):
        nuevo_nodo = nodo(data)
        self.size += 1
        
        if self.head is None:
            self.head = self.tail = self.mitad = nuevo_nodo
        else:
            self.tail.next = nuevo_nodo
            nuevo_nodo.prev = self.tail
            self.tail = nuevo_nodo
            
            # Actualizar mitad sin recorrer la lista
            if self.size % 2 != 0:
                self.mitad = self.mitad.next

    def busqueda_rapida(self, data: int):
        if not self.head: return None
        
        current = self.mitad
        
        # Determinar dirección y buscar sin oscilar
        if current.data == data:
            return current
        
        if current.data < data:
            # Solo buscar hacia adelante
            while current and current.data <= data:
                if current.data == data: return current
                current = current.next
        else:
            # Solo buscar hacia atrás
            while current and current.data >= data:
                if current.data == data: return current
                current = current.prev
                
        return f"el nodo con el dato {data} no existe"
    
    def busqueda_secuencial(self, data: int):
        current = self.head
        while current is not None:
            if current.data == data:
                return current
            current = current.next
        return f"el nodo con el dato {data} no existe"

#! datos de prueba
lista = lista_doble_enlazada()
for i in range(1, 200, 3):
    lista.agregar(i)

print(lista.mitad.data) #? 100
print(lista.busqueda_rapida(61).data) #? 61


````
