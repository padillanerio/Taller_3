Construcción de una Lista Doblemente Enlazada para un Sistema de Gestión de Usuarios: 

Implementa una lista doblemente enlazada donde cada nodo represente un usuario (ID, nombre, email). 

Escribe métodos para añadir, eliminar y buscar usuarios. 

Crea una función para ordenar la lista de usuarios alfabéticamente por nombre y otra por ID. 

```python
class usuario:
    def __init__(self, id, nombre, email):
        self.id = id
        self.nombre = nombre
        self.email = email
```

```python
class node_doble:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None
class ListaEnlazadaDoble:
    def __init__(self):
        self.head = None
        self.tail = None


    def insertar(self,data):
        new_node = node_doble(data)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node

    def modificar(self,old_data,new_data):
        current = self.head
        while current is not None:
            if current.data == old_data:
                current.data = new_data
                return
            current = current.next


    
```

```python
def agregar_usuario(lista: ListaEnlazadaDoble, id_usuario: int, nombre: str, email: str):
    lista.insertar(usuario(id_usuario, nombre, email))


def buscar_usuario_por_id(lista: ListaEnlazadaDoble, id_usuario: int):
    """
    Busca un usuario por su ID.

    Logica:
    - Recorre la lista nodo por nodo.
    - Compara el atributo id del objeto usuario almacenado en cada nodo.
    - Retorna el usuario si lo encuentra o None si no existe.

    Uso:
    - usuario_encontrado = buscar_usuario_por_id(lista_usuarios, 3)
    """
    actual = lista.head
    while actual is not None:
        if actual.data.id == id_usuario:
            return actual.data
        actual = actual.next
    return None


def eliminar_usuario_por_id(lista: ListaEnlazadaDoble, id_usuario: int):
    """
    Elimina un usuario de la lista a partir de su ID.

    Logica:
    - Recorre la lista hasta encontrar el nodo cuyo usuario tenga el ID buscado.
    - Si el nodo es la cabeza, actualiza head.
    - Si el nodo es la cola, actualiza tail.
    - Si el nodo esta en medio, reconecta los nodos anterior y siguiente.

    Uso:
    - eliminar_usuario_por_id(lista_usuarios, 2)
    """
    actual = lista.head

    while actual is not None:
        if actual.data.id == id_usuario:
            if actual == lista.head and actual == lista.tail:
                lista.head = None
                lista.tail = None
            elif actual == lista.head:
                lista.head = actual.next
                lista.head.prev = None
            elif actual == lista.tail:
                lista.tail = actual.prev
                lista.tail.next = None
            else:
                actual.prev.next = actual.next
                actual.next.prev = actual.prev

            actual.next = None
            actual.prev = None
            return True
        actual = actual.next

    return False


# Ejemplo de uso
# agregar_usuario(lista_usuarios, 8, "Hugo", "hugo@example.com")
# print(buscar_usuario_por_id(lista_usuarios, 3).__dict__)
# eliminar_usuario_por_id(lista_usuarios, 2)
```

```python
def clave_nombre(usuario_actual):
    return usuario_actual.nombre.lower()


def clave_id(usuario_actual):
    return usuario_actual.id


def ordenar_por_nombre(lista: ListaEnlazadaDoble):
    """
    Ordena la lista de usuarios alfabeticamente por el atributo nombre.

    Logica:
    - Recorre todos los nodos y extrae sus datos.
    - Ordena la coleccion en memoria por el campo 'nombre'.
    - Reasigna los datos ordenados a los nodos existentes.

    Uso:
    - ordenar_por_nombre(lista_usuarios)
    """
    usuarios = []
    actual = lista.head

    while actual is not None:
        usuarios.append(actual.data)
        actual = actual.next

    usuarios.sort(key=clave_nombre)

    actual = lista.head
    indice = 0
    while actual is not None:
        actual.data = usuarios[indice]
        actual = actual.next
        indice += 1



def ordenar_por_id(lista: ListaEnlazadaDoble):
    """
    Ordena la lista de usuarios de menor a mayor segun el atributo id.

    Logica:
    - Recorre todos los nodos y extrae sus datos.
    - Ordena la coleccion en memoria por el campo 'id'.
    - Reasigna los datos ordenados a los nodos existentes.

    Uso:
    - ordenar_por_id(lista_usuarios)
    """
    usuarios = []
    actual = lista.head

    while actual is not None:
        usuarios.append(actual.data)
        actual = actual.next

    usuarios.sort(key=clave_id)

    actual = lista.head
    indice = 0
    while actual is not None:
        actual.data = usuarios[indice]
        actual = actual.next
        indice += 1


# Ejemplos de uso
# ordenar_por_nombre(lista_usuarios)
# ordenar_por_id(lista_usuarios)
```
