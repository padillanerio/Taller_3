## Implementación de un Historial de Navegación con Listas Doblemente Enlazadas: 

Crea una lista doblemente enlazada para almacenar el historial de navegación de un usuario (cada nodo representa una URL). 

Implementa funciones para moverse hacia adelante y atrás en el historial, y eliminar una URL específica. 

Añade una función para guardar y cargar el historial desde un archivo de texto. 

````python
#! Clase nodo y lista enlazada doble

class node_Url:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None
    
class ListaEnlazadaDoble:
    def __init__(self):
        self.head = None
        self.tail = None
        self.puntero_actual = None


    def insertar(self,data):
        new_node = node_Url(data)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
            self.puntero_actual = self.tail
        else:
            self.tail.next = new_node
            new_node.prev = self.tail
            self.tail = new_node
            self.puntero_actual = self.tail

    def ir_atras(self):
        if self.puntero_actual is not None and self.puntero_actual.prev is not None:
            self.puntero_actual = self.puntero_actual.prev

    def ir_adelante(self):
        if self.puntero_actual is not None and self.puntero_actual.next is not None:
            self.puntero_actual = self.puntero_actual.next

    def buscar_url(self,data):
        current = self.head
        while current is not None:
            if current.data == data:
                return current
            current = current.next
        return None
    
    def eliminar(self,data):
        current = self.buscar_url(data)

        if current is None:
            return f"el nodo con el dato {data} no existe"
        #! Eliminar el nodo encontrado
        #? Si el nodo a eliminar es el head o el tail, se deben actualizar los punteros correspondientes
        if current == self.head and current == self.puntero_actual:
            #? si el nodo a eliminar es el unico nodo en la lista, se deben actualizar ambos punteros head y tail a None
            if self.head == self.tail:
                self.head = None
                self.tail = None
                self.puntero_actual = None
            else:
                #* si el nodo a eliminar es el head pero no es el unico nodo,
                #* se debe actualizar el puntero head al siguiente nodo 
                #* y establecer el puntero prev del nuevo head a None
                self.head = self.head.next
                self.head.prev = None
                current.next = None
                self.puntero_actual = self.head
        elif current == self.tail and current == self.puntero_actual:
            self.tail = self.tail.prev
            self.tail.next = None
            current.prev = None
            self.puntero_actual = self.tail
        else:
            #* si el nodo a eliminar no es ni el head ni el tail, 
            #* se deben actualizar los punteros next y prev de los nodos adyacentes 
            #* para omitir el nodo a eliminar
            current.prev.next = current.next
            current.next.prev = current.prev
            self.puntero_actual = current.prev
            current.next = None
            current.prev = None
    
````

````python
# guardar el histrial de navegacion en un archivo de texto

ruta = "Datos/historial_url.txt"
def guardar_histrial_en_archivo(lista, ruta):
    with open(ruta, "w") as file:
        current = lista.head
        while current is not None:
            file.write(current.data + "\n")
            current = current.next

def cargar_historial_desde_archivo(ruta):
    lista = ListaEnlazadaDoble()
    try:
        with open(ruta, "r") as file:
            for line in file:
                url = line.strip()
                lista.insertar(url)
    except FileNotFoundError:
        print(f"El archivo {ruta} no existe. Se creará un nuevo historial.")
    return lista
````
